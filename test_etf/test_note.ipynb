```python
from pathlib import Path
import os
from dotenv import load_dotenv

ROOT = Path.cwd().resolve()
load_dotenv(ROOT / ".env", override=False)
print("PARQUET_ROOT_PATH=", os.environ.get("PARQUET_ROOT_PATH"))

import panda_data
print("panda_data=", panda_data.__file__)
```

```python
df = panda_data.get_fund_detail(symbol="510050.SH")
print("df shape:", df.shape)
display(df)
```

```python
df2 = panda_data.get_fund_daily(
    start_date="20250101",
    end_date="20250131",
    symbol="510050.SH",
)
print("df2 shape:", df2.shape)
display(df2)
```

```python
assert os.environ.get("PARQUET_ROOT_PATH"), "PARQUET_ROOT_PATH is not loaded from .env"
assert not df.empty, "df is empty"
assert not df2.empty, "df2 is empty"
print("ok")
```
